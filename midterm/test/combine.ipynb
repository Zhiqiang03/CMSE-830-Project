```python
import pandas as pd
import numpy as np
```

```python
green = pd.read_csv('green_taxi.csv', low_memory=False)
yellow = pd.read_csv('yellow_taxi.csv', low_memory=False)
```

```python
print(green.shape)
print(yellow.shape)
```

```python
print(green.columns)
print(yellow.columns)
```

```python
# Define the new names for the green taxi columns
green_rename_map = {
    'lpep_pickup_datetime': 'pickup_datetime',
    'lpep_dropoff_datetime': 'dropoff_datetime',
    'RatecodeID': 'ratecodeid',
    'PULocationID': 'pulocationid',
    'DOLocationID': 'dolocationid'
}

# Apply the renaming
df_green = green.rename(columns=green_rename_map)
```

```python
# Define the new names for the yellow taxi columns
yellow_rename_map = {
    'tpep_pickup_datetime': 'pickup_datetime',
    'tpep_dropoff_datetime': 'dropoff_datetime'
}

# Apply the renaming
df_yellow = yellow.rename(columns=yellow_rename_map)
```

```python
print(df_green.columns)
print(df_yellow.columns)
```

```python
df_combined = pd.concat([df_green, df_yellow], ignore_index=True)
```

```python
df_combined.head()
```

```python
# Fill NaNs for airport_fee with 0 (since green taxis didn't have this fee)
df_combined['airport_fee'] = df_combined['airport_fee'].fillna(0)

# Fill NaNs for trip_type with 1 (assuming yellow taxis are 'street-hail', which is type 1)
df_combined['trip_type'] = df_combined['trip_type'].fillna(1)
```

```python
df_combined.columns
```

```python
df_combined.isna().sum().sort_values(ascending=False)
```

```python
df_combined.to_csv("combined.csv", index=False)
```
