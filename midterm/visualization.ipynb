```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
ride_data = pd.read_csv("./data/ncr_ride_bookings.csv")
fig = px.histogram(
    ride_data,
    x='Booking Status',
    color='Booking Status',  # Colors the bars based on the booking status
    title="Booking Status Distribution (Interactive)",
    labels={'Booking Status': 'Booking Status', 'count': 'Number of Rides'}, # Customize axis labels
    height=450  # Set the height of the plot
)

# Optional: Further customize the layout
fig.update_layout(
    xaxis_title='Booking Status',
    yaxis_title='Number of Rides',
    title_x=0.5, # Center the title
    bargap=0.1 # Add a small gap between bars
)

fig.show()
```

```python
vehicle_counts = ride_data['Vehicle Type'].value_counts().index.tolist()

fig = px.histogram(
    ride_data,
    x='Vehicle Type',
    color='Vehicle Type',  # Colors the bars based on the vehicle type (like 'hue')
    title="Vehicle Type Distribution (Interactive)",
    # Apply the sorted order to the x-axis categories
    category_orders={"Vehicle Type": vehicle_counts},
    labels={'Vehicle Type': 'Vehicle Type', 'count': 'Number of Vehicles'}, # Custom axis labels
    height=450  # Set the height of the plot
)
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
locations = ['Midtown', 'Downtown', 'Airport', 'Brooklyn', 'Queens', 'UES', 'UWS', 'Bronx']
import random
random.seed(42)
ride_data = pd.DataFrame({
    'Pickup Location': random.choices(locations, weights=[10, 8, 7, 5, 4, 3, 2, 1], k=100),
    'Drop Location': random.choices(locations, weights=[1, 2, 3, 4, 5, 7, 8, 10], k=100)
})
# ----------------------------------------------------

# 1. Aggregate data: Get the top 10 locations by count
top_pickups = ride_data['Pickup Location'].value_counts().head(10)
top_drops = ride_data['Drop Location'].value_counts().head(10)

# 2. Create subplots object (1 row, 2 columns)
fig = make_subplots(
    rows=1, cols=2,
    # Set titles for each subplot
    subplot_titles=("Top 10 Pickup Locations", "Top 10 Drop Locations")
)

# 3. Add the first bar chart (Top 10 Pickup Locations)
fig.add_trace(
    go.Bar(
        # Use the counts for the x-axis and the locations for the y-axis
        x=top_pickups.values,
        y=top_pickups.index,
        orientation='h', # Make it a horizontal bar chart
        name='Pickup Count',
        marker_color='#1f77b4'
    ),
    row=1, col=1
)

# 4. Add the second bar chart (Top 10 Drop Locations)
fig.add_trace(
    go.Bar(
        # Use the counts for the x-axis and the locations for the y-axis
        x=top_drops.values,
        y=top_drops.index,
        orientation='h',
        name='Drop Count',
        marker_color='#ff7f0e'
    ),
    row=1, col=2
)

fig.update_yaxes(autorange="reversed")

fig.update_layout(
    title_text="Top 10 Pickup and Drop Locations (Interactive)",
    title_x=0.5, # Center the main title
    height=550, # Set the figure height
    width=1000, # Set the figure width
    # Set axis labels for the subplots
    xaxis1_title='Count',
    xaxis2_title='Count',
)

fig.show()
```
